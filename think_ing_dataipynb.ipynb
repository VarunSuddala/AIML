```python
import json

def load_data(filename):
  with open(filename,"r") as f:
    data=json.load(f)
  return data
```

```python
data = load_data('/content/amazon_store_data.json')
print(type(data))
print(data)
```

Assignment 1

cleaning & structuring our data
_________________________________
1 - data is in consistent

2 - missing data

3 - duplicate data

```python
def clean_data(data):
   text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
   clean_data=[]
   unique=set()

   for user in data:
    # clean rating
      raw_rating=user["rating"]
      if raw_rating in text_to_num:
        user["rating"]=text_to_num[raw_rating]

    #handle missing values (age)

      raw_age=user.get("age")
      if raw_age is not None:
        user["age"]=int(raw_age)
      else :
        user["age"]=None

      #duplicate data(user)
      user_tuple=(user["name"],user["age"])
      if user_tuple not in unique:
        unique.add(user_tuple)
        clean_data.append(user)

   return clean_data
```

```python
cleaned_data=clean_data(data)
for user in cleaned_data:
  print(user)
print()
print(f"length of orginal data {len(data)}")
print(f"length of cleaned data {len(cleaned_data)}")
```

**assignment 2 **

insights from data :

percentage users with poor rating

avg rating

```python
#function to get insights from data
def get_insights(data):
  total_rating=0
  for user in data:
    total_rating+=float(user["rating"])
  avg_rating= total_rating/len(data)
  print(f"average rating {avg_rating}")
  #percentage users with poor rating
  poor_rating_count=0
  for user in data:
    if float(user["rating"])<3.0:
      poor_rating_count+=1
  poor_rating_percentage=(poor_rating_count/len(data))*100
  print(f"percentage of users with poor rating {poor_rating_percentage} %")
```

```python
get_insights(cleaned_data)
```

**Assignment 3**
product recommendation feature

```python
def get_product_recommendations(data):
  recommendations=[]
  curr_rec={}

  for user in data:
    curr_rec["name"]=user["name"]
    if float(user['rating'])>4:
      curr_rec["brand"]="Apple"
    else:
      curr_rec["brand"]="Samsung"
    recommendations.append(curr_rec)
    curr_rec={}
  return recommendations
```

```python
recommendations=get_product_recommendations(cleaned_data)
for rec in recommendations:
  print(rec)
```
